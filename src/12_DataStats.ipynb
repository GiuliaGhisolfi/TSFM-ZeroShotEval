### Stats

```python
import numpy as np
from scipy import stats
from statsmodels.tsa.stattools import acf
from scipy.signal import periodogram


def dominant_period(x, fs=1.0):
    """Trova il periodo dominante (media sulle variabili se multivariata)."""
    x = np.asarray(x, dtype=float)
    if x.ndim == 2:
        return np.nanmean([dominant_period(x[:, i], fs=fs) for i in range(x.shape[1])])
    x = np.nan_to_num(x, nan=0.0)
    n = len(x)
    if n < 4 or np.allclose(x, 0):
        return np.nan
    freqs, power = periodogram(x, fs=fs)
    if len(freqs) < 2:
        return np.nan
    freqs[0] = np.nan
    idx = np.nanargmax(power)
    return np.nan if freqs[idx] == 0 else 1.0 / freqs[idx]


def estimate_trend_slope(x):
    """Trend medio (slope della regressione lineare)."""
    x = np.asarray(x, dtype=float)
    if x.ndim == 2:
        return np.nanmean([estimate_trend_slope(x[:, i]) for i in range(x.shape[1])])
    t = np.arange(len(x))
    if len(x) < 2:
        return np.nan
    slope, _, _, _, _ = stats.linregress(t, np.nan_to_num(x))
    return slope


def autocorr_at_lag(x, lag=1):
    """Autocorrelazione a lag dato."""
    x = np.nan_to_num(x, nan=0.0)
    n = len(x)
    if n <= lag:
        return np.nan
    x_mean = np.mean(x)
    num = np.sum((x[:-lag] - x_mean) * (x[lag:] - x_mean))
    den = np.sum((x - x_mean) ** 2)
    return num / den if den != 0 else np.nan


def snr_estimate(x):
    """Signal-to-noise ratio stimato come var(signal)/var(residual)."""
    x = np.nan_to_num(x, nan=0.0)
    if len(x) < 3:
        return np.nan
    trend = np.polyval(np.polyfit(np.arange(len(x)), x, deg=1), np.arange(len(x)))
    residual = x - trend
    var_signal = np.var(trend)
    var_resid = np.var(residual)
    return np.nan if var_resid == 0 else var_signal / var_resid


def seasonality_strength(x):
    """Stima della forza di stagionalità (ACF-based)."""
    x = np.nan_to_num(x, nan=0.0)
    if len(x) < 3:
        return np.nan
    acfs = acf(x, nlags=min(24, len(x)-1), fft=True)
    acfs = np.nan_to_num(acfs, nan=0.0)
    return np.max(np.abs(acfs[1:]))


def series_stats(series):
    """
    Calcola statistiche generali di una time series (univariate o multivariate).
    """
    x = np.asarray(series, dtype=float)
    if x.ndim == 1:
        n = len(x)
        missing_rate = np.mean(np.isnan(x))
        mean = np.nanmean(x)
        std = np.nanstd(x)
        skew = stats.skew(np.nan_to_num(x))
        kurt = stats.kurtosis(np.nan_to_num(x))
        trend = estimate_trend_slope(x)
        dom_per = dominant_period(x)
        acf1 = autocorr_at_lag(x, 1)
        acf24 = autocorr_at_lag(x, 24) if len(x) > 24 else np.nan
        snr = snr_estimate(x)
        season_strength = seasonality_strength(x)
    else:
        n = x.shape[0]
        missing_rate = np.mean(np.isnan(x))
        mean = np.nanmean(x)
        std = np.nanmean(np.nanstd(x, axis=0))
        skew = np.nanmean([stats.skew(np.nan_to_num(x[:, i])) for i in range(x.shape[1])])
        kurt = np.nanmean([stats.kurtosis(np.nan_to_num(x[:, i])) for i in range(x.shape[1])])
        trend = estimate_trend_slope(x)
        dom_per = dominant_period(x)
        acf1 = np.nanmean([autocorr_at_lag(x[:, i], 1) for i in range(x.shape[1])])
        acf24 = np.nanmean([autocorr_at_lag(x[:, i], 24) for i in range(x.shape[1]) if x.shape[0] > 24])
        snr = np.nanmean([snr_estimate(x[:, i]) for i in range(x.shape[1])])
        season_strength = np.nanmean([seasonality_strength(x[:, i]) for i in range(x.shape[1])])

    return {
        "length": n,
        "missing_rate": missing_rate,
        "mean": mean,
        "std": std,
        "skew": skew,
        "kurtosis": kurt,
        "trend_slope": trend,
        "dominant_period": dom_per,
        "autocorr_lag1": acf1,
        "autocorr_lag24": acf24,
        "snr": snr,
        "seasonality_strength": season_strength,
    }
```

```python
from tqdm import tqdm

def analyze_dataset(df, dataset_name_col="dataset_name", target_col="target"):
    """
    Calcola statistiche per dominio.
    df deve contenere almeno: [domain_col, target_col]
    dove target_col può essere una lista/array di valori.
    """
    results = []
    for dataset, group in tqdm(df.groupby(dataset_name_col)):
        dataset_stats = []
        for _, row in tqdm(group.iterrows()):
            x = row[target_col]
            if isinstance(x, (list, np.ndarray, pd.Series)):
                try:
                    stats_dict = series_stats(x)
                    dataset_stats.append(stats_dict)
                except Exception as e:
                    print(f"Error processing series in dataset {dataset_name_col}: {e}")
                    continue
        if not dataset_stats:
            continue
        dom_df = pd.DataFrame(dataset_stats)
        agg = dom_df.mean(numeric_only=True).to_dict()
        agg["domain"] = dataset
        agg["domain"] = group["domain"].iloc[0] if "domain" in group else "Unknown"
        agg["freq"] = group["freq"].iloc[0] if "freq" in group else "Unknown"
        results.append(agg)
    return pd.DataFrame(results)
```

```python
stats_df = analyze_dataset(all_df, dataset_name_col="dataset_name", target_col="target")
```

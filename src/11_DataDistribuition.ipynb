```python
import joypy
import pandas as pd
from tqdm import tqdm
import numpy as np
import matplotlib.pyplot as plt
```

```python
#domains = all_df["domain"].unique()
domains = ["CloudOps/Web", "Energy", "Econ/Fin", "Nature", "Sales"]
domains
```

```python
def plot_ridgeline_distributions(df, max_samples=1000):
    """
    Ridgeline plot delle distribuzioni per dominio.
    """
    data = []
    for domain in domains:
        # random subset of max_samples series to limit computation
        samples = df[df["domain"] == domain]["target"].to_list()
        if len(samples) > max_samples:
            samples = [samples[idx] for idx in np.random.choice(len(samples), size=max_samples, replace=False)]
        print(f"Processing domain {domain} with {len(samples)} series...")
        for ts in tqdm(samples):
            ts = np.asarray(ts).flatten()
            ts = (ts - np.min(ts)) / (np.max(ts) - np.min(ts)) # min-max scaling
            for val in ts:
                data.append({"value": val, "domain": domain})

    data_df = pd.DataFrame(data)

    joypy.joyplot(
        data_df,
        by="domain",
        column="value",
        figsize=(12, 8),
        colormap=plt.cm.get_cmap("gist_rainbow", len(domains)),
        bins=30,
        fade=True
    )
    plt.title("Ridgeline Plot of Normalized Value Distributions per Domain")
    plt.xlabel("Normalized Time Series Values")
    plt.savefig(f"figures/data_distribuition_per_domain.pdf")
    plt.show()
```

```python
plot_ridgeline_distributions(all_df, max_samples=1000)
```
